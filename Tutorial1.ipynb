```python
import pandas as pd #library for data frame
import numpy as np #library/package for dealing with arrays/numbers
```

1 Create a numpy array of 2 x 2

```python
arr1=np.array([[5,7],[0,9]])
print("This is my numpy 2x2 array")
print(arr1)
```

2 Create a numpy random array of size 3 x 4

```python
#(3,4)the first number is the row and the second is the column
# so this is a grid of 3 rows and 4 columns
# the numbers will be float

rand_arr=np.random.rand(3,4)
print(rand_arr)
```

3 Create a numpy array of ones and a numpy array of zeroes

```python
ones=np.ones((10,5))
zeroes=np.zeros((3,6))
print("Array of ones:")
print(ones)
print("Array of zeroes:")
print(zeroes)
```

4. Create a random numpy array of size 6x3 and transpose it

```python
#when you transpose something you switch the rows and columns
rand_arr=np.random.rand(6,3)
rand_arr_trans=rand_arr.T
print("Original Array(6x3):")
print(rand_arr)
print("Transposed Array(3x6):")
print(rand_arr_trans)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
#one array with n rows and p columns
#another array with p rows and m columns
#dot product

n=3
p=4
m=2

array_A=np.random.rand(n,p)
array_B=np.random.rand(p,m)
dot_product=np.dot(array_A,array_B)

print("Array A:")
print(array_A)
print("Array B:")
print(array_B)
print("This is the dot product:")
print(dot_product)


#second way of doing it
print("Option 2")
print(array_A@array_B)

print("Result shape:",dot_product.shape)
```
